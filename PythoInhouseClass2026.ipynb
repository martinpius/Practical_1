```python
def simulate_student(
        gender: List[AnyStr] = ["male", "female"],program: List[AnyStr] = ["Bsc Math", "Bsc Stat", "Bsc Actuary"],
        grade: List[AnyStr] = ["Pass", "Fail"],
        gpa: List[float] = [1.0, 1.8, 3.0, 4.1],

        )->pd.DataFrame:

        StudId: int = np.arange(start = 0, stop = 100, step = 1)
        gender: AnyStr = random.choices(population = gender, weights = [0.65, 0.35],k = 100)
        program: List[AnyStr] = random.choices(population = program, weights = [0.2, 0.5, 0.3], k = 100)
        gpa = random.choices(population = gpa, weights = [0.1,0.3, 0.4, 0.2],
                             k = 100)
        for g in gpa:
            grade = "PASS" if g > 2.0 else "FAIL"

        stud_data: Dict[AnyStr, object] = {
            "StudId": StudId,
            "gender": gender,
            "degree program": program,
            "gpa": gpa ,
            "grade": grade,
        }
        return pd.DataFrame(data = stud_data)
```

```python
students_df = simulate_student()
```
