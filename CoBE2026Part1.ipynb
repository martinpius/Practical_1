```python
def simulate_normal(
                    x_t: np.ndarray,
                    x_d: np.ndarray,
                    mu: float = 0,
                    sigma: float = 1,
                    density: bool = True,
                    )-> pd.DataFrame:
    """
    -This function simulate a Gaussian distribution

    ---------
    parameters
    ---------
    mu: mean-> scalar float
    sigma: variance-> scalar float
    x_t: vector|numpy array for theoretical distribution
    x_d: vector|numpy array for histogram (emperical distribution)
    density: bool : if True, it plot the pdf curve

    ---------
    Hint: The mathematical eqn of the normal dist [LateX]
    f(x) = (1/(2*pi*sigma**2)**(-0.5)) * exp(-0.5 *(x_t - mu)/sigma)**2. [-inf <= mu, x_t <= inf, sigma > 0]
    """
    k = (2 * np.pi * sigma **2) **-0.5
    E = -0.5*(((x_t - mu)/sigma)**2)

    y = k * np.exp(E)

    # ploting
    plt.figure(figsize = (10, 3))
    plt.hist(x_d,
             bins = 30,
             color = "indigo",
             edgecolor = "black",
             density = density,

             )
    plt.plot(x_t,
             y,
             "r-",
             linewidth = 2.5,
             label = "Theoretical distribution",

             )
    plt.legend(loc = "upper right")
    plt.title(f"The normal pdf with mean {mu} and sigma {sigma}")
    plt.xlabel("x")
    plt.ylabel("prob")
    plt.show()

    data: Dict[str, object] = {"x_t": x_t,
                               "x_d": x_d,
                               "f(x_t)": y}
    return pd.DataFrame(data = data)

```

```python
mu, sigma = 0, 1
x_t = np.linspace(start = mu - 4 * sigma,
                  stop = mu + 4 * sigma,
                  num = 10000)
x_d = np.random.normal(loc = mu, scale = sigma, size = 10000)
```

```python
data = simulate_normal(x_t = x_t,
                       x_d = x_d,
                       mu = mu,
                       sigma = sigma)
```
