```python
# Simulate the dataset for students
# The data must have StudId, Program, Gender, Year, GPA, grade (Pass or Fail), samplesize = 1000
```

```python
random.choices(["A", "B"], weights = [0.8, 0.2], k= 10)
```

```python
def simulate_students(
        gender: List[AnyStr] = ["male", "female"],
        program: List[AnyStr] = ["Bsc. Stat", "B.Sc Maths", "Bsc. Actu", "Bsc. MT"],
        stud_year: List[AnyStr] = [1, 2, 3],
        gpa: List[float] = [1.0, 1.8, 2.1, 3.4, 4.1],
        )-> pd.DataFrame:

        gender: List[AnyStr] = random.choices(
            population = gender,
            weights = [0.65, 0.35], k = 1000,

            )
        program: List[AnyStr] = random.choices(
            population = program,
            weights = [0.2, 0.1, 0.3, 0.4,],
            k = 1000,
        )
        stud_year: List[AnyStr] = random.choices(
            population = stud_year,
            weights = [0.46, 0.23, 0.31,],
            k = 1000,
        )
        stud_id: List[int] = np.arange(start = 100, stop = 100001, step = 100)
        gpa: List[float] = random.choices(
            population= gpa,
            weights = [0.1, 0.15, 0.15, 0.40, 0.2],
            k = 1000,
        )
        for g in gpa:
            grade = "PASS" if g > 2.0 else "FAIL"
        data: Dict[AnyStr, object] = {
            "stud_id":stud_id,
            "gender": gender,
            "program": program,
            "stud_year":stud_year,
            "gpa": gpa,
            "grade": grade,

        }
        return pd.DataFrame(data = data)
```

```python
student_data = simulate_students()
```
