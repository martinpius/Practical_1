```python
def simulate_students(
        gender: List[AnyStr] = ["male", "female"],
        program: List[AnyStr] = ["Bsc. Stat", "B.Sc Maths", "Bsc. Actu", "Bsc. MT"],
        stud_year: List[AnyStr] = [1, 2, 3],
        gpa: List[float] = [1.0, 1.8, 2.1, 3.4, 4.1],
        )-> pd.DataFrame:

        gender: List[AnyStr] = random.choices(
            population = gender,
            weights = [0.65, 0.35], k = 1000,

            )
        program: List[AnyStr] = random.choices(
            population = program,
            weights = [0.2, 0.1, 0.3, 0.4,],
            k = 1000,
        )
        stud_year: List[AnyStr] = random.choices(
            population = stud_year,
            weights = [0.46, 0.23, 0.31,],
            k = 1000,
        )
        stud_id: List[int] = np.arange(start = 100, stop = 100001, step = 100)
        gpa: List[float] = random.choices(
            population= gpa,
            weights = [0.1, 0.15, 0.15, 0.40, 0.2],
            k = 1000,
        )
        for g in gpa:
            grade = "PASS" if g > 2.0 else "FAIL"
        data: Dict[AnyStr, object] = {
            "stud_id":stud_id,
            "gender": gender,
            "program": program,
            "stud_year":stud_year,
            "gpa": gpa,
            "grade": grade,

        }
        return pd.DataFrame(data = data)
```

```python
def simulate_from_norm(
        x_t: np.ndarray,
        x_d: np.ndarray,
        mu: float = 0,
        sigma: float = 1,
        density: bool = True,
        )-> pd.DataFrame:
        """
        -This function simulate the normal distribution embeded with the empirical data

        parameters;
        -----------
        mu: mean--> scalar float (for univariate dist)
        sigma: variance -> scalar float
        x_t: np.ndarray--> vector for computing theoretical pdf
        x_d: np.ndarray-> vector for plolting the empirical dist (histogram)
        density: bool = True, where if True it will plot the PDF

        --------
        [LateX syntax]: X \sim N(mu, sigma) if the pdf is given as
        f(x) = (2*pi*sigma**2)**-0.5 * exp(-0.5*(x - mu)/sigma)**2
        Where -inf < x, mu < inf, sigma > 0
        Hint: Empirical values for x will base on the 4-sigma limits
        """
        # Theoretical distribution
        k = np.sqrt(2 * np.pi * sigma ** 2)**-1
        E = -0.5 * ((x_t - mu)/sigma)**2
        y = k * np.exp(E)

        # Plotting
        plt.figure(figsize = (8, 5))
        plt.hist(x_d,
                 bins = 30,
                 color = "gray",
                 edgecolor = "black",
                 density = density)
        plt.plot(x_t,
                 y,
                 "r-",
                 linewidth = 2.5,
                 label = "Theoretical normal pdf",

                 )
        plt.legend(loc = "upper right")
        plt.ylabel("pdf")
        plt.xlabel("x")
        plt.title(f"The normal density plot with mean {mu} and variance {sigma}")

        data: Dict[AnyStr, np.ndarray] = {
            "x_t": x_t,
            "x_d": x_d,
            "f(x)": y
        }

        return pd.DataFrame(data = data)
```

```python
mu, sigma = 10, 3
x_d = np.random.normal(loc = mu,
                       scale = sigma,
                       size = 10000,
                       )
x_t = np.linspace(start = mu - 4 * sigma,
                  stop = mu + 4 * sigma,
                  num = 10000,
                  )
```

```python
data = simulate_from_norm(x_t = x_t,
                          x_d = x_d,
                          sigma = sigma,
                          mu = mu,)
```

```python
plt.figure(figsize = (6, 6))
plt.scatter(x = data.x_t,
            y = data['f(x)'],
            color = "fuchsia",
            )
plt.title("Scatter plot")
```

```python
# Simulate exponential distribution.
# f(x, theta) = theta * exp(-x * theta), x > 0, theta > 0.
```

```python
def simulate_exp(
        x_t: np.ndarray,
        x_d: np.ndarray,
        theta: float = 1,
        density: bool = True,
        )->pd.DataFrame:

        y = theta * np.exp(-theta * x_t)

        plt.figure(figsize = (10, 4))
        plt.hist(x_d,
                 bins = 30,
                 color = "skyblue",
                 edgecolor = "black",
                 alpha = 0.7,
                 rwidth = 0.75,
                 density = density,
                 label = "eperical distribution"
                 )

        plt.plot(x_t,
                 y,
                 "b-",
                 label = "Theoretical distribution",
                 linewidth = 2.5,
                 )
        plt.title(f"Exponential distribution with theta {theta}")
        plt.ylabel("PDF")
        plt.xlabel("x")
        plt.legend(loc = "best")
        plt.grid(axis = "y",
                 linestyle = "--",
                 alpha = 0.5,)
        plt.show()
        data: Dict[str, object] = {"x_t": x_t, "exp": y}
        return pd.DataFrame(data = data)
```

```python
theta = 3
x_d = np.random.exponential(scale = theta**-1, size = 10000,)
x_t = np.linspace(start = 0,
                  stop = max(x_d), num = 10000,)
```

```python
data = simulate_exp(x_t = x_t, x_d = x_d, theta = theta)
```

```python
def simulate_normscipy(
                       mu: float = 0,
                       sigma: float = 1,
                       density: bool = True,
                       N: int = 1000000,
                       )-> pd.DataFrame:

    x_d: np.ndarray = np.random.normal(
                    loc = mu,
                    scale = sigma,
                    size = N,
                            )

    x_t: np.ndarray = np.linspace(
                     start = mu - 4 * sigma,
                     stop = mu +  4*sigma,
                     num = N,
                     )

    y = norm.pdf(x_t,
                 loc = mu,
                 scale = sigma)

    plt.figure(figsize = (8, 4))
    plt.hist(x_d,
             bins = 30,
             alpha = 0.8,
             color = "skyblue",
             edgecolor = "green",

             density = density,
             label = "Emperical Gaussian",
             )
    plt.plot(x_t,
             y,
            "b-",
             linewidth = 2.5,
             label = "Theoretical Normal Curve"
             )
    plt.legend(loc = "best")
    plt.xlabel("x")
    plt.ylabel("PDF")
    plt.title(f"Normal distribution with mu {mu} and sigma {sigma} ")
    plt.grid(axis = "x",
             linestyle = "--",
             alpha = 0.6)
    plt.show()
    data: Dict[str, object] = {"x_d": x_d, "x_t": x_t, "y": y}
```

```python
data = simulate_normscipy()
```
